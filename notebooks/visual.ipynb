```python
import matplotlib.pyplot as plt 
import pandas as pd
import matplotlib.dates as mdates
import numpy as np
```

```python
df = pd.read_csv(r"D:\pypipeline\data\processed\us_paro_hourly_processed.csv")
```

```python
df.loc[df['pm25']>500,'pm25']=np.nan
```

```python
fig, ax = plt.subplots(figsize=(12,4))
df['date'] = pd.to_datetime(df['date'])
df.index = df['date']

ax.plot(df.loc["2017-3-3":"2018-1-1", 'pm25'])
missing = df['pm25'].isna()

ax.xaxis.set_major_locator(mdates.AutoDateLocator())   # or MonthLocator(6)
ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y"))
ax.fill_between(df.index, df['pm25'], color='skyblue', alpha=0.4,where=missing,step="post")

plt.xlabel("Year")
plt.ylabel("PM2.5")
plt.title("PM2.5 Time Series")
plt.tight_layout()
plt.show()
```
