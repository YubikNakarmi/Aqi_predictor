### Air Quality Index (AQI) Calculation

The AQI is computed for each pollutant using a piecewise linear interpolation
between predefined breakpoint concentrations. The overall AQI is defined as
the maximum of the individual pollutant sub-indices.

$$
I_p = \frac{I_{HI} - I_{LO}}{B_{HI} - B_{LO}} \left( C_p - B_{LO} \right) + I_{LO}
$$

where,

$$
\begin{aligned}
I_p & : \text{AQI sub-index for pollutant } p \\
C_p & : \text{Observed concentration of pollutant } p \\
B_{HI}, B_{LO} & : \text{Upper and lower concentration breakpoints} \\
I_{HI}, I_{LO} & : \text{AQI values corresponding to the breakpoints}
\end{aligned}
$$

The overall AQI is then calculated as:

$$
AQI = \max \left( I_{PM_{2.5}}, I_{PM_{10}}, I_{NO_2}, I_{SO_2}, I_{CO}, I_{O_3} \right)
$$



[Source or Aqi calculation documentation](https://www.airnow.gov/publications/air-quality-index/technical-assistance-document-for-reporting-the-daily-aqi/)
