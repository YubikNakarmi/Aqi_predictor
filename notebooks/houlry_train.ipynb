### Adding horizon targets
training data through horizons and using 24 models to predict each horizon separately.

```python
train_df_cleaned.head()
```

making horizons in the cleaned  data, using segments to avoid leakage

```python
for targets in range(1,25):
    train_df_cleaned[f"pm25_plus_{targets}h"] = (train_df_cleaned.groupby("segment_id")["pm25_target"].shift(-targets))
    train_df_cleaned[f"o3_plus_{targets}h"] = train_df_cleaned.groupby("segment_id")["o3_target"].shift(-targets)
    val_df_cleaned[f"pm25_plus_{targets}h"] = (val_df_cleaned.groupby("segment_id")["pm25_target"].shift(-targets))
    val_df_cleaned[f"o3_plus_{targets}h"] = val_df_cleaned.groupby("segment_id")["o3_target"].shift(-targets)
    test_df_cleaned[f"pm25_plus_{targets}h"] = (test_df_cleaned.groupby("segment_id")["pm25_target"].shift(-targets))
    test_df_cleaned[f"o3_plus_{targets}h"] = test_df_cleaned.groupby("segment_id")["o3_target"].shift(-targets)
```

```python
features_exclude =[f"pm25_plus_{i}h" for i in range(1,25)] + [f"o3_plus_{i}h" for i in range(1,25)] + \
                    ["segment_id",]
#"was_imputed","imputation_confidence","o3_missing","pm25_missing","pm25_gap_length","o3_gap_length"]

models = {}
metrics = {}

```

```python
import xgboost as xgb
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error
params = {"max_depth": 6,
          "learning_rate": 0.1,
          "objective": "reg:absoluteerror",
          "eval_metric": ["mae","rmse"],
          "seed":52,
          "enable_categorical":True
          
          }
```

```python
train_df_cleaned
```

```python
horizon = 24

if "imputation_confidence" in train_df_cleaned.columns and train_df_cleaned["imputation_confidence"].dtype == object:
    train_df_cleaned["imputation_confidence"] = train_df_cleaned["imputation_confidence"].astype("category")

if "imputation_confidence" in val_df_cleaned.columns and val_df_cleaned["imputation_confidence"].dtype == object:
    val_df_cleaned["imputation_confidence"] = val_df_cleaned["imputation_confidence"].astype("category")


for h in range(1,horizon+1):

    train_mask = train_df_cleaned[f"pm25_plus_{h}h"].notnull() & (train_df_cleaned[f"pm25_plus_{h}h"] >=0) \
    & (train_df_cleaned[f"pm25_plus_{h}h"] <=500)

    eval_mask = val_df_cleaned[f"pm25_plus_{h}h"].notnull() & (val_df_cleaned[f"pm25_plus_{h}h"] >=0) \
    & (val_df_cleaned[f"pm25_plus_{h}h"] <=500)

    #using only valid data points
    y = train_df_cleaned.loc[train_mask, f"pm25_plus_{h}h"].reset_index(drop=True).astype(float)
    X = train_df_cleaned.loc[train_mask].drop(columns=features_exclude).reset_index(drop=True)


    y_val = val_df_cleaned.loc[eval_mask, f"pm25_plus_{h}h"].reset_index(drop=True).astype(float)
    X_val = val_df_cleaned.loc[eval_mask].drop(columns=features_exclude).reset_index(drop=True)

    # Skip horizons with insufficient data
    if y.empty or y_val.empty:
        print(f"Skipping horizon {h} due to insufficient data (train={len(y)}, val={len(y_val)})")
        continue

    reg = xgb.XGBRegressor(**params)
    reg.fit(
        X,
        y,
        eval_set=[(X_val, y_val)],
                verbose=False,
    )

    models[f"pm25_plus_{h}h"] = reg
    metrics[f"pm25_plus_{h}h"] = mean_absolute_error(y_val, reg.predict(X_val))
```

```python
# Quick single-feature test to check if a feature is constant and whether model predicts constant values
# Set the horizon and feature to test
h_test = 1
feature_name = 'pm25_lag_2'  # change to any feature column to test

# Build masks and small datasets similar to training loop
train_mask = train_df_cleaned[f"pm25_plus_{h_test}h"].notnull() & (train_df_cleaned[f"pm25_plus_{h_test}h"] >= 0) & (train_df_cleaned[f"pm25_plus_{h_test}h"] <= 500)
val_mask = val_df_cleaned[f"pm25_plus_{h_test}h"].notnull() & (val_df_cleaned[f"pm25_plus_{h_test}h"] >= 0) & (val_df_cleaned[f"pm25_plus_{h_test}h"] <= 500)

# Targets
y_sf = train_df_cleaned.loc[train_mask, f"pm25_plus_{h_test}h"].reset_index(drop=True).astype(float)
X_sf = train_df_cleaned.loc[train_mask, [feature_name]].reset_index(drop=True)

y_val_sf = val_df_cleaned.loc[val_mask, f"pm25_plus_{h_test}h"].reset_index(drop=True).astype(float)
X_val_sf = val_df_cleaned.loc[val_mask, [feature_name]].reset_index(drop=True)

print(f"Single-feature test: horizon={h_test}, feature={feature_name}")
print("Train/Val sizes:", len(y_sf), len(y_val_sf))

if X_sf.empty or X_val_sf.empty:
    print("Insufficient data for this feature/horizon; pick another feature or increase data range.")
else:
    print("Feature train stats:\n", X_sf.describe())
    if X_sf[feature_name].nunique() == 1:
        print(f"WARNING: train feature '{feature_name}' is constant (value={X_sf[feature_name].iloc[0]}).\nModel will likely predict constant.")

    # Simple train
    model_sf = xgb.XGBRegressor(**params)
    model_sf.fit(X_sf, y_sf, eval_set=[(X_val_sf, y_val_sf)], verbose=False)

    preds_sf = model_sf.predict(X_val_sf)
    print("Predictions unique count:", np.unique(preds_sf).shape[0])
    print("Predictions mean/std:", preds_sf.mean(), preds_sf.std())
    print("Validation MAE:", mean_absolute_error(y_val_sf, preds_sf))
    if np.unique(preds_sf).shape[0] == 1:
        print("WARNING: model predictions are constant. Try a different feature, remove over-regularization, or add more informative features.")
```
